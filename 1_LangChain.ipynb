```python
from dotenv import load_dotenv
load_dotenv()

import getpass

import os
from langchain_google_genai import ChatGoogleGenerativeAI

if "GOOGLE_API_KEY" not in os.environ:
  os.environ["GOOGLE_API_KEY"] = getpass.getpass("Enter your Google API")
```

```python
llm = ChatGoogleGenerativeAI(model="gemini-2.5-flash")
```

```python
from langchain_core.messages import SystemMessage, HumanMessage

messages = [
    SystemMessage(content="You are a 5 year old rude kid who helps a user figure out what to eat in one short sentence"),
    HumanMessage(content="what do you like in KFC")
]

reponse = llm.invoke(messages)
print(llm.invoke(messages).content)
```

# LangChain Expression Language

```python
from langchain_core.prompts import ChatPromptTemplate
from langchain_core.output_parsers import StrOutputParser

if "GOOGLE_API_KEY" not in os.environ:
  os.environ["GOOGLE_API_KEY"] = getpass.getpass("Enter your key")

llm = ChatGoogleGenerativeAI(model="gemini-2.5-flash")
template = ChatPromptTemplate.from_template("Tell me a fun fact for {topic}")
parser = StrOutputParser()
```

```python
prompt_value = template.invoke({"topic": "blueberry cheesecake"})
resp_obj = llm.invoke(prompt_value)
print(parser.invoke(resp_obj))
```

```python
chain = template | llm | parser
print(chain.invoke({"topic": "starfish"}))
```

# Assignment

```python
from langchain_core.prompts import ChatPromptTemplate
from langchain_core.output_parsers import StrOutputParser

chain = (
    ChatPromptTemplate.from_template(
        "What is the release year of the movie '{movie}'? "
        "Then calculate how many years ago that was from 2026. "
        "Give a brief answer."
    )
    | llm
    | StrOutputParser()
)

print(chain.invoke({"movie": "Inception"}))
```
